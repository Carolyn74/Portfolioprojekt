## Fragestellungen



### Fragestellungen 1:  Wie haben sich Umsatz und Profitabilität zwischen 2014 und 2017 entwickelt?

```python
entwicklung = df.groupby(
    df["order_date"].dt.year
).agg({
    "gross_sales": "sum",
    "discount_value": "sum",
    "sales": "sum",
    "profit": "sum"
})

entwicklung.round(0)
```

#### Erkenntnis

Nach einem leichten Umsatzrückgang im Jahr 2015 entwickelten sich Umsatz und Gewinn bis 2017 positiv. Gleichzeitig stiegen die gewährten Rabatte deutlich an, was auf einen möglichen Einfluss auf die Profitabilität hindeutet.

### Fragestellung 2: Welchen Einfluss hatten Rabatte auf die Profitabilität?

```python
rabattanalyse = (
    df.groupby(df["order_date"].dt.year)
      .agg({
          "gross_sales": "sum",
          "discount_value": "sum",
          "profit": "sum"
      })
)

rabattanalyse["discount_rate"] = (
    rabattanalyse["discount_value"]
    / rabattanalyse["gross_sales"]
)

rabattanalyse
```

#### Erkenntnis

Die Rabattbelastung lag während des gesamten Betrachtungszeitraums auf einem hohen Niveau. Nach einem Rückgang von 22,2 % im Jahr 2014 auf 18 % im Jahr 2015 blieb sie bis 2017 weitgehend stabil. Obwohl die Rabattquote zurückging, entwickelte sich die Profitabilität nicht proportional zum Umsatzwachstum. Um die zugrunde liegenden Profitabilitätstreiber zu identifizieren, ist daher eine vertiefte Analyse der Rabattstruktur erforderlich.

### Fragestellung 3: Welche Produktgruppen tragen am wenigsten zur Profitabilität bei?

```python
kategorie_analyse = (
    df.groupby("category")
      .agg({
          "gross_sales": "sum",
          "discount_value": "sum",
          "profit": "sum"
      })
)

kategorie_analyse["discount_rate"] = (
    kategorie_analyse["discount_value"]
    / kategorie_analyse["gross_sales"]
)

kategorie_analyse["profit_margin"] = (
    kategorie_analyse["profit"]
    / kategorie_analyse["gross_sales"]
)

kategorie_analyse = (
    kategorie_analyse
      .sort_values(
          by="gross_sales",
          ascending=False
      )
)

kategorie_analyse[["gross_sales", "discount_value", "profit"]] = (
    kategorie_analyse[["gross_sales", "discount_value", "profit"]]
    .round(0)
    .astype(int)
)

kategorie_analyse.round(3)
```

#### Erkenntnis

Die Kategorie Furniture erzielt trotz vergleichbarer Rabattquoten und hoher Umsätze einen deutlich geringeren Gewinnbeitrag als Technology und Office Supplies. Dies deutet darauf hin, dass die schwache Profitabilität nicht allein durch die Höhe der Rabatte erklärt werden kann. Eine vertiefte Analyse auf Subkategorie-Ebene ist daher erforderlich.

### Fragestellung 4: Welche Subkategorien innerhalb der Kategorie Furniture beeinflussen die Profitabilität besonders stark?

```python
subkategorie_analyse["profit_margin"] = (
    subkategorie_analyse["profit"]
    / subkategorie_analyse["gross_sales"]
)

subkategorie_analyse = (
    subkategorie_analyse
        .sort_values(by="gross_sales", ascending=False)
)

subkategorie_analyse[
    ["gross_sales", "discount_value", "profit"]
] = (
    subkategorie_analyse[
        ["gross_sales", "discount_value", "profit"]
    ]
    .round(0)
    .astype(int)
)

subkategorie_analyse.round({
    "discount_rate": 3,
    "profit_margin": 3
})
```

#### Erkenntnis

Die geringe Profitabilität der Kategorie Furniture wird hauptsächlich durch die Subkategorien Tables und Bookcases verursacht. Trotz relevanter Umsätze weisen beide Subkategorien Verluste auf, während Chairs und Furnishings positive Gewinnbeiträge erzielen.

### Fragestellung 5: In welchen Regionen treten die identifizierten Profitabilitätsprobleme besonders häufig auf?

```python
region_analyse = (
    df[df["sub_category"].isin(["Tables", "Bookcases"])]
      .groupby("region")
      .agg({
          "gross_sales": "sum",
          "discount_value": "sum",
          "profit": "sum"
      })
      .sort_values(
          by="profit"
      )
)

region_analyse.round(0).astype(int)
```

#### Erkenntnis

Die identifizierten Profitabilitätsprobleme treten in allen Regionen auf. Besonders betroffen ist die Region East mit dem höchsten Verlustbeitrag, gefolgt von Central und South. Dies deutet darauf hin, dass die geringe Profitabilität von Tables und Bookcases kein regionales Einzelfallproblem, sondern ein strukturelles Problem dieser Subkategorien ist.

### Fragestellung 6: Welche Bundesstaaten tragen am stärksten zu den Verlusten von Tables und Bookcases bei?

```python
staat_analyse = (
    df[df["sub_category"].isin(["Tables", "Bookcases"])]
      .groupby("state")
      .agg({
          "gross_sales": "sum",
          "discount_value": "sum",
          "profit": "sum"
      })
      .sort_values(
          by="profit"
      )
      .head(10)
)

staat_analyse.round(0).astype(int)
```

#### Erkenntnis

Die höchsten Verluste bei Tables und Bookcases konzentrieren sich auf Pennsylvania, Illinois, Texas, New York und Ohio. Diese Bundesstaaten leisten damit den größten Beitrag zu den Profitabilitätsproblemen der beiden Subkategorien.
